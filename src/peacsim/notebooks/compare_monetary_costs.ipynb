```python
from plotting_helpers import get_job_table_df, get_power_trace_df
from peacsim.parsing import policy_config_reader as policy_parser
from peacsim.parsing import job_profile_reader as job_parser
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python

optimization_dir = "W_train_inf_med_util_EDR_1000server_20250312135324/"

weigts_file_path = "../output/optimization/" + optimization_dir + "/opt_weights.csv"   ### mock_weights.csv

experiment =  "Wtrain-inf-short-EDR-1000server_20250312224303" # "W2_NoDR_low_util_1000servers_20250309112952" #  "W2_NoDR_high_util_1000servers_20250309112851" # ""

dr_program = 'EDR'

output_dir = "../output/simulation/" + experiment

policy_config = policy_parser.PolicyConfigReader("../output/optimization/" + optimization_dir + 'optimal_policy.ini')

job_table = get_job_table_df(output_dir)
power_df = get_power_trace_df(output_dir)


experiment_duration_hour = 1

### TRAIN MULTIPPPP
hour = 3600
kwh = hour * 1000
piP, piR, piE, piI = 0.1 / kwh, 0.1 / kwh, 0.1 / kwh, 0.1/kwh
total_power_cost = 0

if dr_program == 'RSR':
    err_phase = power_df['trackingError'][1:]
    abs_err_phase = [abs(float(x)) for x in err_phase.values]
    # P, R = pr_table.at[phase, 'P'], pr_table.at[phase, 'R']

    P, R = policy_config.P, policy_config.R
    power_cost = (piP * P - piR * R) * hour * experiment_duration_hour
    tracking_cost = (piE * R * sum(abs_err_phase) / len(abs_err_phase)) * hour * experiment_duration_hour
    total_power_cost = power_cost + tracking_cost
    print(total_power_cost)
elif dr_program == 'EDR':
    P_max = 711269
    P_avg = 551828
    print("No DR", experiment)

    P= policy_config.P
    print("P:", P)
    total_power_cost = (piP * P - piI * (P_max - P)) * hour * experiment_duration_hour
    print(total_power_cost)
elif dr_program == 'NoDR':
    print("No DR", experiment)
    P = 551828
    total_power_cost = (piP * P) * hour * experiment_duration_hour
    print(total_power_cost)

```
