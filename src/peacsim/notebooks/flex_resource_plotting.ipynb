```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
# Data
app_names = ['Resnet.infer.4', 'GPT2.infer.4', 'Llama.infer.4', 'Bloom.infer.4']
percentage_of_time_qos_satisfied = np.asarray([0.763677, 0.811441, 0.739239, 0.774507]) * 100
qos_requirement = np.asarray([0.8, 0.8, 0.8, 0.8]) * 100

# Create a DataFrame for plotting
df = pd.DataFrame({
    "Application": app_names,
    "QoS Satisfied": percentage_of_time_qos_satisfied,
    "QoS Requirement": qos_requirement
})

# Set Seaborn style
sns.set_style("whitegrid")
plt.rcParams.update({
    "font.size": 14,    # General font size
    "axes.labelsize": 16,  # Axis labels
    "axes.titlesize": 18,  # Title size
    "xtick.labelsize": 12, # X-ticks (smaller to prevent overlap)
    "ytick.labelsize": 14, # Y-ticks
    "legend.fontsize": 14  # Larger legend size
})

# Create bar plot
plt.figure(figsize=(8, 5))
ax = sns.barplot(x="Application", y="QoS Satisfied", data=df, color="skyblue", alpha=0.8)

# Add QoS Requirement line
plt.plot(df["Application"], df["QoS Requirement"], linestyle="--", color="red", marker="o", markersize=6, label="QoS Requirement")

# Labels and title
plt.xlabel("Applications", fontsize=14)
plt.ylabel("Percentage of Time QoS Satisfied", fontsize=14)
# plt.title("QoS Satisfaction vs Requirement", fontsize=16)

# Add legend
plt.legend(fontsize=12)

plt.savefig("qos_satisfaction_vs_requirement.png")
# Show plot
plt.show()
```
